```python
# Assignment Operators
a = 10
b = 6
c = 70.2
place = "Perungudi"

# Set Initialization (duplicates removed automatically)
numbers = {1, 1, 2, 2, 2, 3, 4}

# List Initialization
location = ["Chennai", "Madurai", "Delhi"]

# Tuple Initialization
company = ("IBM", "Infosys", "CTS", "TCS")

# Dictionary Initialization
detail = {"Name": "Pradeepraj", "Age": 41, "City": "Chennai"}

# Arithmetic Operators
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)

# Comparison Operators
print("Is a equal to b:", a == b)
print("Is a not equal to b:", a != b)
print("Is a greater than b:", a > b)
print("Is a lesser than b:", a < b)

# Logical Operators
print(a > 2 and a < 11)
print(a > 2 or a < 11)
print(not (a > 2))

# Assignment Operations
print(a)
a += 5
print(a)
a -= 2
print(a)

# Type Checking
print(a, type(a))
print(c, type(c))
print(place, type(place))
print(a, type(a > b))

# Set Operations
print(numbers)
numbers.add(5)
print(numbers)

# List Operations
print(location)
location.append("Banglore")
print(location[2])
print(location, location[2])

# Tuple Operations
print(company)
print(company, company[1])

# Dictionary Operations
print(detail["Name"], detail["Age"])
detail["Phone"] = 9884552537
detail["Blood"] = "O+"
print(detail)

# Control Flow
if a >= b:
    print("A is greater than or equal to B")
else:
    print("A is lesser than or equal to B")

if a + b >= c:
    print("Sum of A and B is greater than C")
elif a + b == c:
    print("Sum of A and B is equal to C")
elif a + b <= c:
    print("Sum of A and B is less than C")
else:
    print("C is greater than the sum of A and B")

# Final Print (Note: variable 'd' is not defined)
# print("Value of A is", a, "Value of B is", b, "Value of C is", c, "Value of D is", d)
```
