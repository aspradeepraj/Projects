# 🚀 Python Backend Study Guide: Introduction to FastAPI

FastAPI is a modern, high-performance web framework designed for building production-grade backend services and APIs with Python. This guide covers installation workflows, application initialization, local server execution, and API validation via Postman.

---

## 📥 Core Framework Installation

FastAPI requires an Asynchronous Server Gateway Interface (ASGI) production server engine to execute. Use the package manager inside your activated virtual environment (`venv`) to deploy both dependencies:

```bash
pip install fastapi
pip install uvicorn
```

---

## 📝 Creating a Basic API (`main.py`)

Create a fresh python script file named `main.py` and implement the baseline structure below to instantiate your server entry boundaries:

```python
from fastapi import FastAPI

# Initializes the core FastAPI application object engine instance
app = FastAPI()

# Decorator defining an HTTP GET operation route at the root URL path ("/")
@app.get("/")
def read_root():
    # Returns a structured Python dictionary which FastAPI automatically serializes into JSON text
    return {"message": "Hello World"}
```

### Architectural Component Breakdown
* **`app = FastAPI()`**: Initializes the root web application ecosystem.
* **`@app.get("/")`**: A routing decorator mapping HTTP GET requests on the root path directly to the underlying function logic.
* **`read_root()`**: The execution block housing the functional response logic triggered upon endpoint matching.

---

## ⚡ Running the Local Development Server

Launch the web application container process via your terminal window using the Uvicorn engine execution script:

```bash
uvicorn main:app --reload
```

* **`main`**: References the target Python file name context matching your codebase script (`main.py`).
* **`app`**: Points directly to the initialized FastAPI application instance variable inside that file.
* **`--reload`**: Enactivates an ongoing file watcher loop that automatically reboots the execution thread upon code modifications.

### 🌐 Network Address Access Mapping
* **Default Endpoint**: `http://127.0.0.1:8000` (localhost)
* **Custom Port Redirection**: To spin up the backend container infrastructure across non-default port windows, pass the port argument flag explicitly:
  ```bash
  uvicorn main:app --reload --port 5000
  ```

## 🛣️ Understanding Endpoints & HTTP Methods

Endpoints are designated destination URLs exposed by your backend interface architecture. They map structural operations straight to target CRUD database actions:

| HTTP Method | Core Operational Objective | Equivalent Database Action (CRUD) |
| :--- | :--- | :--- |
| **`GET`** | Retrieve data assets from the environment. | Read |
| **`POST`** | Inject entirely new data payloads. | Create |
| **`PUT`** | Modify or overwrite an existing data element. | Update |
| **`DELETE`** | Purge or drop a record permanently. | Delete |

### 🛠️ Working with Dynamic Path Parameters
You can implement path routing structures to dynamically capture unique query records straight from the incoming URL:

```python
@app.get("/items/{item_id}")
def read_item(item_id: int):
    # Enforces static type checking validation ensuring 'item_id' matches an integer parameter
    return {"item_id": item_id}
```

# 🧮 Python Backend Study Guide: API Implementation & Core Concepts

## 🛠️ Setting Up the Calculator API (`calculator_api.py`)

A Calculator API provides web endpoints to process mathematical operations. Rather than forcing users to execute Python scripts locally, they send parameters over HTTP and receive calculated results instantly.

Save this complete, self-contained implementation as `calculator_api.py` inside your project directory:

```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel

# Initialize the backend API engine
app = FastAPI(
    title="Automated Calculator API",
    description="Production-grade API endpoints for basic arithmetic operations"
)

# Define the structure for incoming JSON payloads using a Pydantic Model
class CalculationRequest(BaseModel):
    operation: str  # Options: 'add', 'subtract', 'multiply', 'divide'
    num1: float
    num2: float

@app.post("/calculate")
def calculate(request: CalculationRequest):
    # Extracts parameters from the verified incoming JSON request body
    op = request.operation.lower().strip()
    n1 = request.num1
    n2 = request.num2

    # Execute corresponding execution branch logic based on the operational string
    if op == "add":
        result = n1 + n2
    elif op == "subtract":
        result = n1 - n2
    elif op == "multiply":
        result = n1 * n2
    elif op == "divide":
        # Handle structural zero division constraints cleanly
        if n2 == 0:
            raise HTTPException(status_code=400, detail="Math Error: Cannot divide by zero.")
        result = n1 / n2
    else:
        raise HTTPException(status_code=400, detail=f"Unsupported operation: '{op}'")

    # Returns the payload as a dictionary which FastAPI serializes to clean JSON text
    return {
        "operation": op,
        "inputs": {"num1": n1, "num2": n2},
        "result": result
    }
```

## ⚡ Running the API & Automated Documentation

Start your calculator application instance by executing this command inside your terminal:

```bash
uvicorn calculator_api:app --reload
```

To compute numbers using your live API, transmit a `POST` network request to the `/calculate` path. The payload body must match the standard structured JSON formatting rules.
